# Stream API

#### Example

```python
import time
import os
from eventregistry import EventRegistry, GetRecentArticles

er = EventRegistry(apiKey=os.environ["NEWSAPI_AI_KEY"])

feed = GetRecentArticles(er)

# Poll the stream in a loop (new articles ~every minute)
while True:
    articles = feed.getUpdates()
    print(f"{len(articles)} new articles")
    for article in articles:
        print(article["uri"], "|", article["title"])
    time.sleep(60)  # wait a minute before next poll
```
